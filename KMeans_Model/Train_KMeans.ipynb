```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import joblib

# 1. Load the training data from your project
df = pd.read_csv("train_data.csv")

print("Columns in train_data.csv:")
print(df.columns.tolist())

# In this dataset, genre one-hot columns start at index 8 (after Medium)
genre_cols = df.columns[8:]
genre_df = df[genre_cols]

print(f"\nNumber of genre features: {len(genre_cols)}")

# 2. Simulate users based on game genres
def simulate_users(genre_df, num_users=200, games_per_user=10):
    """
    Create fake 'users' by randomly sampling games and averaging their genre vectors.
    Each user vector = mean of the genres of N random games.
    """
    users = []

    for _ in range(num_users):
        sample_games = genre_df.sample(games_per_user, replace=True)
        user_profile = sample_games.mean().values  # average genre distribution
        users.append(user_profile)

    return np.array(users)

X = simulate_users(genre_df)

print("User feature matrix shape:", X.shape)

# 3. Train K-Means clustering
k = 4
model = KMeans(n_clusters=k, random_state=42)
labels = model.fit_predict(X)

# 4. Evaluate with silhouette score
score = silhouette_score(X, labels)
print(f"Silhouette Score for k={k}: {score:.3f}")

# 5. Save the model and data in Colab
joblib.dump(model, "kmeans_model.pkl")
np.save("user_vectors.npy", X)
np.save("cluster_labels.npy", labels)

print("Training complete — saved kmeans_model.pkl, user_vectors.npy, cluster_labels.npy")
```
